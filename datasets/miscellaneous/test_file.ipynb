```python
import json

# Load JSONL
with open("train.jsonl", "r", encoding="utf-8") as f1:
    data1 = [json.loads(line) for line in f1]

# Load normal JSON
with open("clustered_math2000.json", "r", encoding="utf-8") as f2:
    data2 = json.load(f2)
```

```python
# Build lookup dictionary
problem_to_example = {
    example["problem"]: example
    for example in data1
}

data3 = []

for item in data2:
    problem = item["messages"][0]["content"]
    if problem in problem_to_example:
        data3.append(problem_to_example[problem])
```

```python
with open("test.jsonl", "w", encoding="utf-8") as f:
    for ex in data3:
        f.write(json.dumps(ex) + "\n")
```
