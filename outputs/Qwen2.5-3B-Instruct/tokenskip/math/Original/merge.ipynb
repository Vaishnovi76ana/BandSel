```python
import json

files = [
    "train_500/predictions.jsonl", "train_1000/predictions.jsonl", "train_1500/predictions.jsonl", "train_2000/predictions.jsonl", "train_2500/predictions.jsonl",
    "train_3000/predictions.jsonl", "train_3500/predictions.jsonl", "train_4000/predictions.jsonl", "train_4500/predictions.jsonl", "train_5000/predictions.jsonl",
    "train_5500/predictions.jsonl", "train_6000/predictions.jsonl", "train_6500/predictions.jsonl", "train_7000/predictions.jsonl", "train_7500/predictions.jsonl"
]

with open("predictions.jsonl", "w") as outfile:
    for fname in files:
        with open(fname, "r") as infile:
            for line in infile:
                if line.strip():         # skip empty lines
                    json.loads(line)     # validate JSON
                    outfile.write(line)
```

```python
import json
import glob

metric_files = glob.glob("train_*/metrics.json")

final = {
    "n_samples": 0,
    "accuracy": 0.0,
    "avg_cot_length": 0.0,
    "sample_latency": 0.0,
}

accuracies = []
cot_lengths = []
latencies = []

for fname in metric_files:
    with open(fname, "r") as f:
        m = json.load(f)
        final["n_samples"] += m["n_samples"]
        accuracies.append(m["accuracy"])
        cot_lengths.append(m["avg_cot_length"])
        latencies.append(m["sample_latency"])

final["accuracy"] = sum(accuracies) / len(accuracies)
final["avg_cot_length"] = sum(cot_lengths) / len(cot_lengths)
final["sample_latency"] = sum(latencies) / len(latencies)

with open("final_metrics.json", "w") as f:
    json.dump(final, f, indent=4)

print("Collected", len(metric_files), "metrics files")
print(json.dumps(final, indent=4))
```
