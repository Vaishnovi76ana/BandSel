```python
import json

data = []

file_path = "samples/predictions.jsonl"
with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():          
            data.append(json.loads(line))
```

```python
unique_datasets = set()

for example in data:
    unique_datasets.add(example["type"])

print(unique_datasets)
```

```python
print((data[0]["accuracy"]))
```

```python
data_list = {}

for example in unique_datasets:
    data_list[example] = [0,0,0]

for example in data:
    key = example["type"]

    if example["accuracy"] == True:
        data_list[key][0] += 1
    else:
        data_list[key][1] += 1
    data_list[key][2] +=1

print(data_list)
```
